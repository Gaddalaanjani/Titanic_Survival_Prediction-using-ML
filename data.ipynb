# Feature Engineering

```python
def feature_engineering(data):
    data=data.copy()
    
    #Extract passenger title
    data["Title"]=(data["Name"].str.extract(r",\s*([^.]*)\.",expand=False).str.strip())
    
    #combine family members
    data["FamilySize"]=data["SibSp"]+data["Parch"]+1

    #identity passenger travelling alone
    data["IsAlone"]=(data["FamilySize"]==1).astype(int)
    data["Has_Cabin"] = (
        data["Cabin"].notna()
    ).astype(int)
    #remove unneccessary columns
    data.drop(columns=["PassengerId","Name","Ticket","Cabin"], inplace=True)


    return data 

```

```python
data=feature_engineering(df)
```

```python
data.head()
```

```python
data[["Title","FamilySize","IsAlone"]].head()
```

```python
x=data.drop("Survived",axis=1)

y=data["Survived"]
```

```python
print("Features:", x.shape)
print("Target:", y.shape)
```

# Train- test split

we divide the dataset into :

1. 80% trainig data
2. 20% testing data

```python
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# Save model
import joblib

import warnings
warnings.filterwarnings("ignore")
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=42,stratify=y)
```

```python
print("Trainig data",x_train.shape)
print("testing data",x_test.shape)
```

# Handle missing values and encode categorial data

```python
numeric_features=x_train.select_dtypes(include=["int64","float64"]).columns.tolist()
```

```python
numeric_features
```

```python
categorial_feature=x_train.select_dtypes(include=["object"]).columns.tolist()
```

```python
categorial_feature
```

```python
print("numerical features",numeric_features)
print("categorial features",categorial_feature)
```

# create preprocessing pipeline

```python
numeric_transformer=Pipeline(steps=[("imputer",SimpleImputer(strategy="median")),
                                    ("scaler",StandardScaler())])
```

```python
numeric_transformer
```

```python
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])
```

```python
categorical_transformer
```

```python
preprocessor=ColumnTransformer(transformers=[
    ("num",numeric_transformer,numeric_features),
    ("cat",categorical_transformer,categorial_feature)
])
```

```python
preprocessor
```

This automatically handles missing values, scales numerical columns and encodes categorical columns using the training data.

# Bulid Machine learning models

# create models

```python
models={
    "LogisticRegression":LogisticRegression(max_iter=1000),

    "Decision Tree":DecisionTreeClassifier(random_state=42),

    "Random Forest":RandomForestClassifier(
        n_estimators=200,
        random_state=42
    )
}
```

```python
models
```

# Train all models

```python
trained_models={}
```

```python
trained_models={}
for name, model in models.items():
    
    pipeline = Pipeline(steps=[
        ("preprocessor",preprocessor),
        ("classifier",model)
    ])

    pipeline.fit(x_train, y_train)

    trained_models[name] = pipeline

    print(name, "Training completed")
```

# model Evaluation

```python
result=[]

for name, model in trained_models.items():

    y_pred=model.predict(x_test)
    y_prob=model.predict_proba(x_test)[:,-1]
    accuracy=accuracy_score(y_test, y_pred)
    precision=precision_score(y_test, y_pred)
    recall=recall_score(y_test,y_pred)
    f1=f1_score(y_test,y_pred)
    roc_auc=roc_auc_score(y_test,y_prob)

    result.append({
        "model":name,
        "Accuracy":accuracy,
        "precision":precision,
        "Recall":recall,
        "f1 score":f1,
        "Roc Auc":roc_auc
    })
result_df=pd.DataFrame(result)
```

```python
print(result_df)
```

```python
print(result_df.round(4))
```

# compare model accuracy

```python
plt.Figure(figsize=(8,5))

sns.barplot(
    data=result_df,
    x="model",
    y="Accuracy"
)
plt.title("Machine Learning Model Comparison")
plt.xticks(rotation=15)
plt.ylim(0, 1)

plt.show()
```

# classification report

```python
best_model_name=result_df.sort_values(
    by="f1 score",
    ascending=False
).iloc[0]["model"]

best_model=trained_models[best_model_name]
y_pred=best_model.predict(x_test)
print("selected model :",best_model_name)
print(classification_report(y_test, y_pred))
```

# confusion matrix

```python
cm=confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True,fmt="d",cmap="Blues",xticklabels=["Not Survived","survived"],yticklabels=["Not survived","survived"])

plt.xlabel("predicted")
plt.ylabel("Actual")

plt.title("confusion matrix")
plt.show()
```

# cross-validation

```python
for name, model in trained_models.items():
    scores=cross_val_score(model, x_train, y_train, cv=5, scoring="accuracy")
    print(name)
    print("cv scores",scores)
    print("mean Accuracy",scores.mean())
    print("------------------------------")
```
